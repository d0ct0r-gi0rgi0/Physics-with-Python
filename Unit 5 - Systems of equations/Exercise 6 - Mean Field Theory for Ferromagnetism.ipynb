## Exercise 6: Mean Field Theory for Ferromagnetism

In mean field theory of ferromagnetism, the magnetization intensity $M$ of a ferromagnetic material depends on temperature according to the equation:

$$M=\mu\tanh\frac{J\,M}{k_B\,T},$$

where $μ$ is the magnetic moment, $J$ is a coupling constant, and $k_B$ is the Boltzmann constant.

Defining $m=M/\mu$ and $C= \mu J/k_B$, the magnetization per unit magnetic moment is given by:

$$m=\tanh\frac{C\,m}{T}.$$

Taking $C=1$, find the solution of the equation beyond the trivial solution, $m=0$, for temperatures $0 <T< 2$ K with a tolerance for the solution $\epsilon =10^{-6}$.


```python
def f(m, T):
    C = 1
    return np.tanh(C*m/T)

m = np.linspace(0, 5, 1000)
plt.plot(m, f(m, 1), color = 'mediumorchid')
plt.grid()
plt.title("Magnetization for T = 1 and C = 1")
plt.show()
```

```python
def fixedpoint(f, T, eps, printing = True):
    it = 0
    err = 1.0
    x0 = 1.0
    while err > eps:
        x1 = f(x0, T)
        err = abs(x1 - x0)
        it += 1
        x0 = x1
    if printing:
        print("Fixed-Point method's iterations:", it)
    return x1

T = np.linspace(0, 2, 1000)
sol = [fixedpoint(f, t, 1e-6, False) for t in T]

plt.plot(T, sol, color = 'mediumorchid')
plt.grid()
plt.title('Solution to m(T) = $tanh(\\frac{Cm}{T})$')
plt.xlabel('T (k)')
plt.ylabel('m')
plt.show()
```
