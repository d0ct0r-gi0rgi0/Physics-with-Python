### Ejercicio 3: propagación de epidemias.

La ecuación $x=1-\text{e}^{-cx}$ aparece en modelos de procesos de contacto, de epidemias y en la teoría de gráficos aleatorios.

1. Escribir un programa que resuelva la ecuación utilizando el método del punto fijo para c = 2. Calcular la solución con una precisión de al menos $10^{-6}$.


2. Modificar el programa para calcular la solución para valores de c de 0 a 3 en pasos de 0.01 y hacer un plot de x como función de c. 


3. ¿Existe una transición de Fase?. (Es la llamada transición de percolación o el umbral epidémico). 

```python
def f(x, c):
    return 1 - np.exp(-c * x)


def fixedpoint(f, c, eps):
    x0 = 1.0
    err = 1.0
    while err > eps:
        x1 = f(x0, c)
        err = abs(x1 - x0)
        x0 = x1
    return x0


a = fixedpoint(f, 2.0, 1e-6)
print(a)

cons = np.linspace(0.0, 3.0, 300)
sol = [fixedpoint(f, c, 1e-6) for c in cons]

plt.plot(cons, sol, color = 'mediumorchid')
plt.grid()
plt.title(f"Solution to $x = 1 - e^{'{-cx}'}$")
plt.xlabel('c')
plt.ylabel('x')
plt.show()
```
