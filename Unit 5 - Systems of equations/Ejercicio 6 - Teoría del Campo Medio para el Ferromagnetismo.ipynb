## Ejercicio 6:  teoría del campo medio para el ferromagnetismo

En la teoría de campo medio del ferromagnetismo, la intensidad de magnetización $M$ de un material ferromagnético depende da la temperatura según la ecuación:

$$M=\mu\tanh\frac{J\,M}{k_B\,T},$$

donde $μ$ es el momento magnético, $J$ es una constante de acoplo y $k_B$ es la constante de Boltzmann.

Definiendo $m=M/\mu$ y $C= \mu J/k_B$ la magnetización por unidad de momento magnético viene dada:

$$m=\tanh\frac{C\,m}{T}.$$

Cogiendo $C=1$, encontrar la solución de la ecuación más allá de la trivial, m=0, para temperaturas $0 <T< 2$ K con una tolerancia para la solución $\epsilon =10^{-6}$.

```python
def f(m, T):
    C = 1
    return np.tanh(C*m/T)

m = np.linspace(0, 5, 1000)
plt.plot(m, f(m, 1), color = 'mediumorchid')
plt.grid()
plt.title("Magnetization for T = 1 and C = 1")
plt.show()
```

```python
def fixedpoint(f, T, eps, printing = True):
    it = 0
    err = 1.0
    x0 = 1.0
    while err > eps:
        x1 = f(x0, T)
        err = abs(x1 - x0)
        it += 1
        x0 = x1
    if printing:
        print("Fixed-Point method's iterations:", it)
    return x1

T = np.linspace(0, 2, 1000)
sol = [fixedpoint(f, t, 1e-6, False) for t in T]

plt.plot(T, sol, color = 'mediumorchid')
plt.grid()
plt.title('Solution to m(T) = $tanh(\\frac{Cm}{T})$')
plt.xlabel('T (k)')
plt.ylabel('m')
plt.show()
```
